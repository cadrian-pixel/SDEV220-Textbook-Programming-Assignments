```python
import zoo
print(zoo.hours())

import zoo as menagerie
print(menagerie.hours())

import sqlite3
conn = sqlite3.connect('books.db')
cursor = conn.cursor()
cursor.execute('''CREATE TABLE IF NOT EXISTS books (title TEXT, author TEXT, year INTEGER)''')

import csv
books2 = [
    ['The Weirdstone of Brisingamen', 'Alan Garner', 1960],
    ['Perdido Street Station', 'China Miéville', 2000],
    ['Thud!', 'Terry Pratchett', 2005],
    ['The Spellman Files', 'Lisa Lutz', 2007],
    ['Small Gods', 'Terry Pratchett', 1992]
]
with open('books2.csv', 'wt') as file:
    writer = csv.writer(file)
    writer.writerow(['title', 'author', 'year'])
    writer.writerows(books2)
with open('books2.csv', 'rt') as file:
    reader = csv.reader(file)
    next(reader)
    for row in sorted(reader):
        if row == []:
            continue
        else:
            print(row)
conn.commit()
conn.close()

import sqlalchemy as sal
engine = sal.create_engine('sqlite:///books.db')
print(f"Engine: {engine}")
```
