```python
SECRET = 7
GUESS = 5
if GUESS < SECRET:
    print("too low")
elif GUESS > SECRET:
    print("too high")
else:
    print("just right")
SMALL = True
GREEN = False
if SMALL and GREEN:
    print("pea")
elif SMALL and not GREEN:
    print("cherry")
elif not SMALL and GREEN:
    print("watermelon")
else:
    print("pumpkin")
ALIST = [3, 2, 1, 0]
for value in ALIST:
    print(value)
GUESS_ME = 7
NUMBER = 1
while True:
    if NUMBER < GUESS_ME:
        print("too low")
    elif NUMBER == GUESS_ME:
        print("found it!")
        break
    else:
        print("oops")
        break
    NUMBER += 1
GUESS_ME = 5
for NUMBER in range(10):
    if NUMBER < GUESS_ME:
        print("too low")
    elif NUMBER == GUESS_ME:
        print("found it!")
        break
    else:
        print("oops")
        break
```
